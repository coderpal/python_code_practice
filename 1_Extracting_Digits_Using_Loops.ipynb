```python
### How will we do that?

## Suppose n is the given number

## we will assign a variable "num" that is equal to "n"
# num = n

## STEP 01: we will use "while" loop where num > 0
# while num > 0

## STEP 02: we will use modulo to extract or mark the last digit in the number
# last_digit = num % 10

## STEP 03:  we will find subsequent numbers using floor division, dividing by 10
# num // 10
```

### 1.1 Count Number of Digits in a given integer

```python
def count_of_digits(n):

  # lets begin the count with 0
  count = 0

  # lets begin with while loop, where num > 0
  while n > 0:

    # increment the count by 1 at every instance
    count += 1

    # floor division by 10
    n = n // 10

  # return count variable after it reaches 0 and the loop stops
  return count

# call the functions with various examples
count_of_digits(245678)

count_of_digits(5873)
```

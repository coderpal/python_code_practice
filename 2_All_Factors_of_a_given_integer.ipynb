## TOPIC: Find all factors of a given integer

```python
def allFactors(num):

  # create an empty list
  result = []

  # utilize for loop
  # should go only upto half the integer (mathematical formula)
  for n in range(1, num//2 + 1):
    if num % n == 0:
      result.append(n)
  result.append(num)    # since when we divide that integer by half, we wont be having the given integer itself

  return result
```

```python
# check by calling the function
allFactors(10)
```
